# 故事生成器智能体演示

## 项目介绍
这是一个支持多种AI模型的故事生成器，可以生成小说、诗歌、剧本等多种文本内容。

### 支持的模型
- 🤖 智谱AI GLM-4.5
- 🔷 OpenAI GPT系列
- 🔷 Anthropic Claude
- 🔷 Google Gemini
- 🔷 其他兼容OpenAI接口的模型

### 功能特性
- ✅ 小说生成（支持短篇、中篇、长篇）
- ✅ 诗歌生成（支持多种形式）
- ✅ 剧本生成（支持多种类型）
- ✅ 内容总结
- ✅ 文本翻译
- ✅ 自动API检测和适配

```python
# 1. 环境准备
import sys
import os
import warnings
warnings.filterwarnings('ignore')

# 添加项目路径
current_dir = os.path.dirname(os.path.abspath('__file__'))
if current_dir not in sys.path:
    sys.path.insert(0, current_dir)

print(f"当前目录: {current_dir}")
print("✅ 环境准备完成")
```

## 2. 检查配置

```python
# 加载环境变量
try:
    from dotenv import load_dotenv
    load_dotenv()
    config_loaded = True
except ImportError:
    config_loaded = False

# 显示配置信息
print("=== 配置信息 ===")
print(f"环境变量加载: {'✅ 成功' if config_loaded else '❌ 失败'}")
print(f"API密钥: {'✅ 已设置' if os.getenv('OPENAI_API_KEY') else '❌ 未设置'}")
print(f"基础URL: {os.getenv('BASE_URL', 'OpenAI')}")
print(f"模型名称: {os.getenv('MODEL_NAME', 'gpt-4')}")
print(f"温度: {os.getenv('TEMPERATURE', '0.7')}")

# 检查依赖
required_packages = ['openai', 'requests', 'pydantic']
missing_packages = []

for package in required_packages:
    try:
        __import__(package)
    except ImportError:
        missing_packages.append(package)

if missing_packages:
    print(f"\n⚠️ 缺少依赖: {', '.join(missing_packages)}")
    print("请运行: pip install openai requests pydantic")
else:
    print("\n✅ 所有必要依赖已安装")
```

## 3. 初始化智能体

```python
# 导入并初始化故事生成器
try:
    from src.agent import StoryGeneratorAgent
    
    # 创建智能体实例
    print("正在初始化故事生成器...")
    agent = StoryGeneratorAgent()
    
    # 获取API信息
    api_info = agent.llm_client.get_api_info()
    
    print("\n✅ 故事生成器初始化成功！")
    print(f"API类型: {api_info['api_type']}")
    print(f"模型名称: {api_info['model_name']}")
    print(f"基础URL: {api_info['base_url'] or 'OpenAI官方接口'}")
    
except Exception as e:
    print(f"❌ 初始化失败: {e}")
    import traceback
    traceback.print_exc()
```

## 4. 小说生成演示

```python
# 生成小说
print("\n=== 小说生成 ===\n")

novel_params = {
    "theme": "一个关于友谊的故事",
    "style": "现实主义",
    "length": "短篇"
}

print(f"生成参数: {novel_params}")
print("\n正在生成...")

try:
    novel_content = agent.generate_novel(**novel_params)
    print("\n--- 生成的小说 ---\n")
    print(novel_content)
except Exception as e:
    print(f"❌ 生成失败: {e}")
```

## 5. 诗歌生成演示

```python
# 生成诗歌
print("\n=== 诗歌生成 ===\n")

poem_params = {
    "theme": "春天",
    "style": "抒情诗",
    "form": "十四行诗"
}

print(f"生成参数: {poem_params}")
print("\n正在生成...")

try:
    poem_content = agent.generate_poem(**poem_params)
    print("\n--- 生成的诗歌 ---\n")
    print(poem_content)
except Exception as e:
    print(f"❌ 生成失败: {e}")
```

## 6. 剧本生成演示

```python
# 生成剧本
print("\n=== 剧本生成 ===\n")

script_params = {
    "theme": "一个悬疑故事",
    "style": "现代剧",
    "genre": "悬疑",
    "scene_count": 2
}

print(f"生成参数: {script_params}")
print("\n正在生成...")

try:
    script_content = agent.generate_script(**script_params)
    print("\n--- 生成的剧本 ---\n")
    print(script_content)
except Exception as e:
    print(f"❌ 生成失败: {e}")
```

## 7. 内容总结演示

```python
# 内容总结
print("\n=== 内容总结 ===\n")

test_text = "人工智能技术正在快速发展，它已经深入到我们生活的各个方面。从智能手机的语音助手到自动驾驶汽车，AI正在改变我们的生活方式。同时，AI也面临着伦理和安全方面的挑战，如何在技术发展的同时确保人类的安全和隐私，是我们需要认真思考的问题。"

print(f"原始文本长度: {len(test_text)} 字符")
print("\n原始文本:\n")
print(test_text)
print("\n正在总结...")

try:
    summary = agent.summarize(test_text)
    print("\n--- 总结结果 ---\n")
    print(summary)
except Exception as e:
    print(f"❌ 总结失败: {e}")
```

## 8. 文本翻译演示

```python
# 文本翻译
print("\n=== 文本翻译 ===\n")

translate_params = {
    "text": "The story generator is a powerful tool that can create various types of content.",
    "language": "中文"
}

print(f"翻译文本: {translate_params['text']}")
print(f"目标语言: {translate_params['language']}")
print("\n正在翻译...")

try:
    translation = agent.translate(translate_params['text'], translate_params['language'])
    print("\n--- 翻译结果 ---\n")
    print(translation)
except Exception as e:
    print(f"❌ 翻译失败: {e}")
```

## 9. 交互式生成（可选）

```python
# 交互式生成功能
def interactive_demo():
    """交互式演示"""
    print("\n=== 交互式生成器 ===\n")
    
    print("支持的生成类型:")
    print("1. novel - 小说")
    print("2. poem - 诗歌")
    print("3. script - 剧本")
    
    try:
        generation_type = input("\n请选择生成类型 (1/2/3): ")
        
        if generation_type == '1':
            theme = input("请输入小说主题: ")
            style = input("请输入风格: ")
            length = input("请输入长度 (短篇/中篇/长篇): ")
            print("\n正在生成小说...")
            content = agent.generate_novel(theme, style, length)
            
        elif generation_type == '2':
            theme = input("请输入诗歌主题: ")
            style = input("请输入风格: ")
            form = input("请输入形式 (自由诗/格律诗/十四行诗): ")
            print("\n正在生成诗歌...")
            content = agent.generate_poem(theme, style, form)
            
        elif generation_type == '3':
            theme = input("请输入剧本主题: ")
            style = input("请输入风格: ")
            genre = input("请输入类型 (喜剧/悲剧/科幻/悬疑): ")
            scene_count = int(input("请输入场景数量: "))
            print("\n正在生成剧本...")
            content = agent.generate_script(theme, style, genre, scene_count)
        
        else:
            print("无效选择")
            return
        
        print("\n--- 生成结果 ---\n")
        print(content)
        
    except Exception as e:
        print(f"❌ 生成失败: {e}")

# 取消注释以下行来运行交互式演示
# interactive_demo()
```
