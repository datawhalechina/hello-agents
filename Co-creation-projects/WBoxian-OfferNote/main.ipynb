Hello-Agents 毕业设计入口。默认用虚构材料和临时数据库运行真实 API 闭环；离线演示不输出能力评分，无需密钥。演示反馈是固定规则，不能作为真实模型质量指标。

先在本项目目录安装 `requirements.txt`；Web 启动和真实模型配置见 README。

```python
from pathlib import Path
import sys

candidates = [Path.cwd(), *Path.cwd().parents]
for parent in list(candidates):
    co_creation = parent / 'Co-creation-projects'
    if co_creation.is_dir():
        candidates.extend(co_creation.glob('*-OfferNote'))
roots = [p for p in candidates if (p / 'scripts/demo.py').is_file() and (p / 'backend/app/main.py').is_file()]
if not roots:
    raise RuntimeError('请从 OfferNote 项目目录启动 Jupyter，或切换到项目目录后重新运行。')
ROOT = roots[0].resolve()
sys.path.insert(0, str(ROOT))
print('项目入口已就绪')
```
