# StudyBuddyAgent

## 项目简介
基于 HelloAgents 框架的学习伙伴智能体，支持知识点讲解、代码点评和出题练习。

## 第1部分：环境配置

````python
# 安装依赖（如果需要）
# !pip install -q -r requirements.txt

from hello_agents import SimpleAgent, HelloAgentsLLM
from hello_agents.tools import BaseTool
from dotenv import load_dotenv

# 加载环境变量
load_dotenv()

print("✅ 环境配置完成")
````

## 第2部分：工具定义

````python
class CustomTool(BaseTool):
    """自定义工具类"""

    name = "tool_name"
    description = "工具描述"

    def run(self, query: str) -> str:
        """工具执行逻辑"""
        return f"处理结果：{query}"

print("✅ 工具定义完成")
````

## 第3部分：智能体构建

````python
llm = HelloAgentsLLM()

system_prompt = """你是一个耐心的学习伙伴 StudyBuddy。

你的任务是：
1. 用通俗易懂的语言讲解知识点
2. 点评学习者的代码并给出改进建议
3. 根据学习内容出题，检验学习效果
"""

agent = SimpleAgent(
    name="StudyBuddy",
    llm=llm,
    system_prompt=system_prompt
)

agent.add_tool(CustomTool())

print("✅ 智能体构建完成")
````

## 第4部分：功能演示

````python
# 示例1：知识点讲解
print("=== 示例1：知识点讲解 ===")
result = agent.run("用通俗的语言解释一下什么是递归")
print(result)
````

````python
# 示例2：代码点评
print("=== 示例2：代码点评 ===")
with open("data/sample_code.py", encoding="utf-8") as f:
    code = f.read()
result = agent.run(f"请点评以下代码并给出改进建议：\n```python\n{code}\n```")
print(result)
````
