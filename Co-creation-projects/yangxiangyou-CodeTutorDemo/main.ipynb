````python
import sys
from pathlib import Path

project_root = Path.cwd()
src_path = project_root / 'src'
if str(src_path) not in sys.path:
    sys.path.insert(0, str(src_path))
````

````python
from tutor.graph import build_tutor_graph
from tutor.model import FakeLLMAdapter
from tutor.state import init_state
````

## 示例一：编程问答

````python
question_llm = FakeLLMAdapter([
    '{"answer": "for 循环适合遍历已有序列，while 循环适合在条件满足时反复执行。"}',
    '{"practice": "请你分别写一个 for 和 while 输出 1 到 5 的小程序。"}',
])
question_graph = build_tutor_graph(question_llm).compile()
question_state = question_graph.invoke(init_state(user_input='Python 里的 for 循环和 while 循环有什么区别？'))
print(question_state['summary'])
````

## 示例二：代码点评

````python
review_llm = FakeLLMAdapter([
    '{"answer": "这段代码的主要问题是函数定义后少了冒号，因此会报语法错误。"}',
    '{"practice": "请你补全冒号后，再写一个 subtract 函数练习。"}',
])
review_graph = build_tutor_graph(review_llm).compile()
review_input = '请帮我看看\n```python\ndef add(a, b)\n    return a + b\n```'
review_state = review_graph.invoke(init_state(user_input=review_input))
print(review_state['summary'])
````
