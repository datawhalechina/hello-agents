## 第3部分：工具定义

HelloAgents框架的核心理念是「万物皆为工具」。我们定义三个核心工具：

1. **WebSearchTool** - 网络搜索工具
2. **TextSummarizerTool** - 文本摘要工具
3. **ReportGeneratorTool** - 报告生成工具

```python
# 导入自定义工具
from src.tools import WebSearchTool, TextSummarizerTool, ReportGeneratorTool

# 创建工具实例
web_search = WebSearchTool()
summarizer = TextSummarizerTool()
report_gen = ReportGeneratorTool()

print("✅ 工具加载完成")
print(f"   - {web_search.name}: {web_search.description}")
print(f"   - {summarizer.name}: {summarizer.description}")
print(f"   - {report_gen.name}: {report_gen.description}")
```

## 第4部分：构建智能研究助手

使用HelloAgents的SimpleAgent和工具系统，构建完整的智能研究助手。

```python
from hello_agents import SimpleAgent, HelloAgentsLLM, ToolRegistry

# 创建LLM实例
llm = HelloAgentsLLM()

# 创建工具注册表
tool_registry = ToolRegistry()
tool_registry.register_tool(web_search)
tool_registry.register_tool(summarizer)
tool_registry.register_tool(report_gen)

# 定义系统提示词
system_prompt = """你是一个专业的智能研究助手。你的任务是帮助用户进行深度研究。

你拥有以下工具：
1. web_search - 搜索互联网获取最新信息
2. text_summarizer - 将长文本压缩为摘要
3. report_generator - 生成结构化的研究报告

当需要使用工具时，请使用以下格式：
[TOOL_CALL:工具名:参数]

例如：
- 搜索：[TOOL_CALL:web_search:人工智能最新进展]
- 摘要：[TOOL_CALL:text_summarizer:这里是要摘要的文本]
- 生成报告：[TOOL_CALL:report_generator:主题|发现内容|来源]

请始终保持客观、准确，引用可靠的来源。"""

# 创建智能体
agent = SimpleAgent(
    name="SmartResearchAgent",
    llm=llm,
    system_prompt=system_prompt,
    tool_registry=tool_registry,
    enable_tool_calling=True
)

print("✅ SmartResearchAgent 创建完成")
print(f"   Agent名称: {agent.name}")
print(f"   可用工具: {agent.list_tools()}")
```

### 示例2：文本摘要

测试文本摘要功能。

```python
# 测试文本摘要工具
print("=== 测试2：文本摘要 ===")
sample_text = """
大语言模型（Large Language Model, LLM）是近年来人工智能领域最重要的突破之一。
这些模型通过在海量文本数据上进行预训练，获得了强大的语言理解和生成能力。
2025年，Agent技术成为AI领域的热点，各大公司纷纷推出自己的Agent框架。
研究表明，基于LLM的Agent能够自主完成复杂任务，包括代码编写、数据分析、
科学研究等。关键创新包括ReAct范式、工具调用、记忆系统等。
然而，Agent技术仍面临幻觉问题、推理效率、安全性等挑战。
未来发展方向包括多模态Agent、协作Agent系统、以及更高效的推理方法。
"""

summary = summarizer.run({"text": sample_text, "max_points": 3})
print(summary)
```

### 示例3：完整研究流程

现在演示完整的研究流程：搜索 → 分析 → 生成报告。

```python
# 使用Agent进行完整研究
print("=== 示例3：智能研究 ===")
print("\n🔬 开始研究：大语言模型Agent技术\n")

# 让Agent执行研究任务
research_result = agent.run("请帮我研究'大语言模型Agent技术'的最新进展，包括主要框架、应用场景和未来趋势。请使用搜索工具收集信息，然后生成一份研究报告。")

print("\n" + "="*60)
print("📋 研究结果：")
print("="*60)
print(research_result)
```

### 示例4：生成结构化报告

使用报告生成工具创建正式的研究报告。

```python
# 生成结构化报告
print("=== 示例4：生成研究报告 ===")

# 准备研究发现
findings = """
1. Agent框架百花齐放：LangChain、AutoGen、CAMEL等框架持续演进
2. ReAct范式成为主流：结合推理和行动的Agent设计模式
3. 工具调用能力增强：Agent能够调用各种外部工具和服务
4. 多Agent协作：多个Agent协同完成复杂任务
5. 记忆系统创新：长期记忆和短期记忆的结合
"""

sources = """
- 来源1：arXiv论文 - Agent技术综述
- 来源2：GitHub热门Agent项目
- 来源3：AI技术博客和新闻
"""

report = report_gen.run({
    "topic": "大语言模型Agent技术",
    "findings": findings,
    "sources": sources
})

print(report)
```

### 示例5：保存报告到文件

```python
# 保存报告到文件
import os

output_dir = "outputs"
os.makedirs(output_dir, exist_ok=True)

report_path = os.path.join(output_dir, "research_report.md")
with open(report_path, "w", encoding="utf-8") as f:
    f.write(report)

print(f"✅ 报告已保存到: {report_path}")
```
