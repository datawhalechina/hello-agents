基本配置

```python
import os
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI

load_dotenv()

llm = ChatOpenAI(
    model=os.getenv("MODEL"),
    temperature=0,
    openai_api_key=os.getenv("API_KEY"),
    openai_api_base=os.getenv("BASE_URL")
)
from agents.main_agent import MAIN_TOOLS, MAIN_SYSTEM_PROMPT
```

```python
from langchain.agents import create_agent
# 【关键修改】使用 create_react_agent，且参数是 prompt
agent_executor = create_agent(llm, MAIN_TOOLS, system_prompt=MAIN_SYSTEM_PROMPT)

# 测试复杂指令
response = agent_executor.invoke({"messages": [("user", "帮我查一下 600519 现在的行情，顺便看看它最近技术面怎么样，MACD金叉了吗？")]})
print(response["messages"][-1].content)
```
